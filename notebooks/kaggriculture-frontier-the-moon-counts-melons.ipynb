# Kaggriculture Frontier | The Moon Counts Melons



**V135: live-public Moon refinement.** V135 preserves V132's complete dynamic route and changes only one dependency-complete early carrot seed order. When the public farms are still structurally mirrored, it keeps one bootstrap carrot seed and relies on the existing just-in-time replenishment already performed after every realized carrot plant. All other states return V132 exactly.


```python
from pathlib import Path
import base64, hashlib, io, tarfile, zlib
import matplotlib.pyplot as plt
import pandas as pd
from kaggle_environments import make
import kaggle_environments

CANDIDATE_SHA = '61d233a41682f026ac81be2538afc621fd3057af21ef7b4582fc1513e70f8e63'
LOCAL_EVIDENCE_ENGINE_VERSION = '1.32.7'
RUNTIME_ENGINE_VERSION = kaggle_environments.__version__
assert RUNTIME_ENGINE_VERSION == LOCAL_EVIDENCE_ENGINE_VERSION
print('Candidate SHA:', CANDIDATE_SHA)
print('Runtime/evidence engine:', RUNTIME_ENGINE_VERSION)
```

## Strategy

V135 retains V132's delayed opening, current/legacy layouts, public shop livestock routes, carrot relay, bounded tomato and strawberry modules, capacity guards, terminal controller, and Brunch/Smoothie second-Yarn fallback. The new gate is evaluated at the existing carrot order and uses only public farm shapes: hands, quadrants, structures, crops, and animals. It reads no player identity, episode seed, geometry fingerprint, reward, future shop, or private rival state.

When active, the smaller bootstrap order releases early cash for wheat and animal feed while the inherited route's existing one-seed replenishment keeps every actual carrot planting supplied. When inactive, the exact V132 action object is returned.


```python
fit = pd.DataFrame([
    ['V127 public stream', 74, 55, 19, 14586.2],
    ['V132 public stream', 34, 32, 2, 27570.7],
    ['V127 public opponents (distinct)', 44, 38, 6, 23360.2],
    ['Inherited current-top routes', 42, 41, 1, 16379.8],
], columns=['panel','games','wins','losses','mean_margin'])
display(fit)
ax=fit.set_index('panel')[['wins','losses']].plot.bar(
    color=['#c99a39','#5865A6'], figsize=(10,4), rot=10,
    title='Frozen V135 official-engine evidence')
ax.set_ylabel('games'); plt.tight_layout()
```

## Frozen selection result

V135 was evaluated against every locally captured opponent from the two currently climbing Moon submissions. On the 74-row V127 public stream it kept the 55-19 outcome record while improving mean margin by 307.3 over V132: 42 rows improved, 28 were exact, and four already-decided rows regressed by at most 391. On the 34-row V132 stream it improved the record from 30-4 to 32-2 in exact local counterfactuals.

The refinement fixes a real cash dependency observed in public losses: V132 ordered eight early carrot seeds although only five carrots were planted and the route already replenished after each plant. The released cash restores otherwise-failed wheat/feed orders. A universal version was rejected after a severe held-out regression; the public mirror gate is the bounded replacement.


```python
comparison = pd.DataFrame([
    ['V127 stream (74)', '55-19', '55-19', 42, 28, 4, 307.3],
    ['V132 stream (34)', '30-4', '32-2', 20, 12, 2, 197.2],
    ['Distinct public 44', '38-6', '38-6', 20, 22, 2, 230.5],
], columns=['evidence','V132','V135','improved','exact','regressed','mean_delta'])
display(comparison)
```

## Live-public mechanism

In the Kirito and HyperX public-loss routes, the original eight-seed order left eight carrot seeds in stock after every later planting because the inherited relay bought a replacement seed each time. Reducing only the bootstrap quantity made an early wheat purchase and sheep feed affordable, recovered wool output, and improved both seats. The mirror-state condition excludes the Victor counterexample that invalidated the universal change.


```python
hard = pd.DataFrame([
    ['Wufang', '10-10', '10-10', 42.9, 42.9],
    ['Kuroda', '19-1', '19-1', 5098.7, 5098.7],
    ['Fabian', '0-20', '0-20', -10855.4, -10855.4],
    ['Pirhosseinlou', '20-0', '20-0', 15684.7, 15684.7],
    ['Tyooughtul', '20-0', '20-0', 6070.5, 6070.5],
], columns=['family','V107 record','V113 record','V107 mean','V113 mean'])
display(hard)
```

## Honest boundary

V135 is a leaderboard candidate, not a claim that one route beats every opponent. It does not repair the first V132 public loss to Konno: that game has no stranded output and is a whole-opening/market interaction. The separate Soil route wins that exact local matchup and is therefore being deployed independently rather than forced into an invalid cross-state Moon splice. Public rating remains the decisive test.

All comparisons used official Python engine 1.32.7, frozen candidate hashes, fresh spawned episodes, and both seats. V135 accepts small already-winning margin regressions in exchange for higher mean and two recovered losses on the freshest stream; only the tied public submission can measure the population result.

## Live execution proof

Two deterministic episodes exercise the exact materialized V135 agent through all 720 calls in both seats using engine 1.32.7. This is the runtime/package gate before the tied notebook creates the public competition submission.


```python
runs=[]
for seat,seed in [(0,62010),(1,62011)]:
    agents=[str(MAIN_PATH),'starter'] if seat==0 else ['starter',str(MAIN_PATH)]
    env=make('kaggriculture',configuration={'episodeSteps':720,'seed':seed},debug=False)
    env.run(agents); mine=env.state[seat]
    assert mine.status=='DONE',(seat,mine.status)
    runs.append({'runtime_engine':RUNTIME_ENGINE_VERSION,'seat':seat,'seed':seed,'status':mine.status,'reward':mine.reward})
display(pd.DataFrame(runs))
```

## Root-only bundle

The final code cell writes one archive member, `main.py`, and proves its bytes equal the displayed candidate hash.


```python
ARCHIVE=Path('submission.tar.gz')
with tarfile.open(ARCHIVE,'w:gz') as tf:
 info=tarfile.TarInfo('main.py'); info.size=len(candidate_bytes); info.mtime=0; info.mode=0o644
 tf.addfile(info,io.BytesIO(candidate_bytes))
with tarfile.open(ARCHIVE,'r:gz') as tf:
 names=tf.getnames(); bundled=tf.extractfile('main.py').read()
assert names==['main.py'] and bundled==candidate_bytes
print('Bundle verified:',hashlib.sha256(ARCHIVE.read_bytes()).hexdigest())
```

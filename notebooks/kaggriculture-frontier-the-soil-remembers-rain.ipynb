# Kaggriculture Frontier | The Soil Remembers Rain

**V202A: closed-loop Moon frontier with two causal suffix repairs.**

V202A preserves the completed 2683.1-rated V197 architecture and changes policy only after two exact public decision forks that caused observed losses. Each repair hands the remaining game to a prefix-compatible V174 controller. No opponent identity, rating, replay id, episode seed, future shop, reward, or private rival state is used.


```python
import base64, hashlib, importlib.metadata, tarfile, warnings, zlib
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-whitegrid')
CANDIDATE_SHA = 'ccd38e0ce94059796176adde117a6a6041fd9422ba70aa445d3b972be02c697a'
EVIDENCE_ENGINE = '1.32.7'
RUNTIME_ENGINE = importlib.metadata.version('kaggle-environments')
assert RUNTIME_ENGINE == EVIDENCE_ENGINE
print('Candidate SHA:', CANDIDATE_SHA)
print('Runtime/evidence engine:', RUNTIME_ENGINE)
```

## Strategy

V202A inherits V197's V174 opening and lawful shop router. After the first public shop, `SMOOTHIE_SHOP`, `YARN_STORE`, and `FARMERS_MARKET` may use the observation-driven Kaito planner; `PET_CAFE` to `YARN_STORE` and `BRUNCH_SPOT` to `YARN_STORE` are the two public second-shop routes already delegated by V197.

The first repair targets first-shop Yarn only. At the exact shared feed action, if the public farm has at most eight planted Wheat tiles, Kaito would feed without buying Wheat, and V174 would execute the same movement while buying the four missing Wheat units. V202A takes that V174 action and keeps the compatible V174 suffix for the rest of that seat's game.

The second repair targets Brunch-to-Yarn only. At the exact shared Wool/Cow fork, it hands off to V174 when own public cash is at least 365 and rival public cash is at most 500. The fork additionally requires the complete expected actor, hand, and market actions from both controllers. Once selected, the handoff is sticky so no later incompatible graft is possible.


```python
validation = pd.DataFrame([
    ['V197 public-loss corpus', 32, 1, 31, -10151.78, -24595],
    ['V202A public-loss corpus', 32, 6, 26, -7379.53, -24595],
    ['Frozen current top: V201E', 200, 181, 19, 21110.825, -26121],
    ['Frozen current top: V202A', 200, 181, 19, 21110.825, -26121],
    ['Four new V189C losses: V202A', 8, 8, 0, 27786.38, 3748],
], columns=['panel','games','wins','losses','mean_margin','worst_margin'])
display(validation)
ax=validation.set_index('panel')[['wins','losses']].plot.bar(
    color=['#5D8C63','#B84A62'], figsize=(10,4), rot=12,
    title='Official-engine V202A admission gates')
ax.set_ylabel('games'); plt.tight_layout()
```

## Frozen selection result

All sixteen completed V197 public losses were downloaded and converted into execution-guarded public-route opponents. Under each original seed and both candidate seats, V202A improved the direct record from V197's 1-31 to 6-26, raised mean margin by 2,772.25, recovered three source episodes in both seats, and preserved the -24,595 worst tail.

The anti-regression gate then reran V202A on 200 exact current-top cells: ten frozen public opponents, ten held-out seeds, and both seats. V202A finished 181-19 with mean +21,110.825 and worst -26,121. Every one of the 200 paired margins equaled V201E's recorded result. Rewards, public terminal hashes, both public farm hashes, private-state hashes, completion statuses, and turn counts were also identical on all 200 cells.


```python
comparison = pd.DataFrame([
    ['Public-loss rows', 32, 5, 26, 1, 5, 0, 2772.25],
    ['Frozen-top paired rows', 200, 0, 200, 0, 0, 0, 0.0],
], columns=['evidence','rows','candidate_higher','exact_margin','baseline_higher',
            'net_new_wins','new_losses','mean_margin_delta'])
display(comparison)
```

## Causal public-loss mechanisms

Two first-Yarn losses shared Kaito's feed-without-Wheat fork but reached it with only six or eight planted Wheat tiles. A one-action edit repaired only one episode; the sticky compatible suffix repaired both episodes in both seats. The selector is therefore based on the public farm deficiency plus exact current actions, not the source replay.

The Brunch-to-Yarn loss reached an exact Wool/Cow fork with own cash 368 and rival cash 351. V174's compatible suffix changed the Wool sale from 8 to 9 plus one later unit and won both seats. Seven strong-field comparison states, including an old sentinel recovered by V197, stayed outside the own-cash/rival-cash regime. The complete 200-cell top gate subsequently proved no collision on the frozen population.


```python
architecture = pd.DataFrame([
    ['Base opening', 'steps 0-71', 'exact V174 controller'],
    ['Public shop router', 'revealed first/second shops', 'bounded Kaito delegation'],
    ['Low-Wheat Yarn repair', 'public Wheat <= 8 plus exact feed fork', 'sticky V174 suffix'],
    ['Brunch-Yarn liquidity repair', 'own cash >=365, rival cash <=500, exact fork', 'sticky V174 suffix'],
    ['Fallback', 'all other observations', 'exact inherited V197/V201E action'],
], columns=['layer','lawful evidence','controller'])
display(architecture)
```

## Honest boundary

V202A does not solve every V197 loss. The newest five-hand, one-pasture, $2,498 opening against Kaileh57 still loses the exact guarded rerun by 2,099 in both seats. V198B, V189C, and V200I can win that seed, but all three diverge from V197 at action zero, when every lawful observation is identical. A universal opening replacement would therefore require a separate broad architecture gate and is not smuggled into this suffix-only candidate.

The public score remains decisive. V197 completed at 2683.1 after 83 games; V189C fell to 2142.1 and is rejected as a deployment architecture despite strong local pre-submission proxies.


```python
evidence = pd.DataFrame([
    ['candidate_sha256','ccd38e0ce94059796176adde117a6a6041fd9422ba70aa445d3b972be02c697a'],
    ['engine','1.32.7'],
    ['public_control','V197: 2683.1 after 83 games; 66-16-1'],
    ['public_loss_gate','V202A 6-26 vs V197 1-31; same -24,595 worst'],
    ['public_loss_mean','V202A -7,379.53 vs V197 -10,151.78'],
    ['frozen_top_200','181-19; mean +21,110.825; worst -26,121'],
    ['frozen_top_pairing','200/200 equal margins; 0 trajectory-field differences'],
    ['new_v189c_losses','8-0; mean +27,786.38; worst +3,748'],
    ['selectors','first-Yarn low Wheat; Brunch-Yarn two-sided liquidity'],
    ['competition_submission','pending this public Soil notebook version'],
], columns=['field','value'])
display(evidence)
```

## Why this composition survived

Both repairs occur after an exact common prefix and at a decision where the candidate can verify the complete current action pair. Both are sticky, so the selected controller owns every downstream dependency: feed, farm movement, livestock, production accounting, liquidity, and sales. The intervention is narrow enough to be falsifiable, but it is a complete compatible suffix rather than replay-specific action memorization.


```python
rejections = pd.DataFrame([
    ['One-action low-Wheat patch', 'repaired the six-Wheat episode but not the eight-Wheat suffix'],
    ['V189C delayed opening', 'public submission regressed to 2142.1; 1-7 on four new losses'],
    ['V198B universal opening', 'wins the newest V197 loss but 1-7 on four new V189C losses'],
    ['V148 family', '0-32 on its exact frozen-top loss cells'],
    ['Action-zero route by future shop', 'future shop unavailable when the opening must diverge'],
    ['Identity/rating/replay/seed routing', 'unavailable, non-causal, and prohibited'],
], columns=['rejected intervention','reason'])
display(rejections)
```

## Live execution proof

The exact materialized `main.py` plays complete official-engine episodes from both seats with Kaggle Environments 1.32.7. This verifies the public payload and runtime before the notebook creates its competition artifact.


```python
from kaggle_environments import make

live_rows=[]
for side,seed in [(0,70110),(1,70111)]:
    agents=[str(MAIN_PATH),'starter'] if side==0 else ['starter',str(MAIN_PATH)]
    env=make('kaggriculture',configuration={'episodeSteps':720,'seed':seed},debug=False)
    env.run(agents)
    mine=env.state[side]
    assert mine.status=='DONE',(side,mine.status)
    live_rows.append({'candidate seat':side,'seed':seed,'status':mine.status,'reward':mine.reward})
display(pd.DataFrame(live_rows))
```

## Root-only package integrity

The final code cell creates `submission.tar.gz` with exactly one root member, `main.py`, reopens it, and proves the member bytes equal the displayed V202A SHA-256. The archive is generated by this public notebook rather than uploaded as an opaque local source.


```python
BUNDLE_PATH = Path("submission.tar.gz")
with tarfile.open(BUNDLE_PATH, "w:gz") as bundle:
    bundle.add(MAIN_PATH, arcname="main.py")

with tarfile.open(BUNDLE_PATH, "r:gz") as bundle:
    names = bundle.getnames()
    packed = bundle.extractfile("main.py").read()

assert names == ["main.py"]
assert hashlib.sha256(packed).hexdigest() == EXPECTED_AGENT_SHA256
print("bundle members", names)
print("bundle sha256", hashlib.sha256(BUNDLE_PATH.read_bytes()).hexdigest())
```
